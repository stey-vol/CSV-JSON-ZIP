## 2. Модуль `csv`
CSV расшифровывается как **Comma-Separated Values**, то есть "значения, разделенные разделителем".

CSV часто используют для таблиц. Например:
```
name, age, city
Иван, 20, Москва
Анна, 22, Казань
Петр, 19, Минск
```
Первая строка содержит названия столбцов, а следующие строки - данные.

Хотя в названии формата упоминается запятая, в реальных файлах разделителем может быть `;`
, табуляция или другой символ. Поэтому в Python разделитель можно указать явно.
## Зачем нужен `csv`
Модуль полезен, когда программа должна:
- прочитать таблицу из файла
- записать результаты в таблицу
- обработать список строк без ручного разбора текста
- обменяться табличными данными с другой программой

Например, программа может получить выгрузку продаж в CSV, прочитать ее и посчитать статистику.


````python
import csv
````

## 2.1. Создание CSV-файла
Для записи таблицы используем `csv.writer`.

`writerow()` записывает одну строку, а `writerows()` - несколько строк.

Параметр `newline=''` нужен при работа с CSV, чтобы избежать появления лишних пустых строк на некоторых системах.

````python
import csv
students=[
    ['Иван', 20, 'Москва'],
    ['Анна', 22, 'Казань'],
    ['Петр', 19, 'Минск']
]
with open('students.csv', 'w', encoding='utf-8', newline='') as file:
  writer=csv.writer(file)
  writer.writerow(['name','age','city'])
  writer.writerows(students)
print('Файл students.csv создан')
````

### Что произошло в этом коде
1. `open(...,'w',...)` открыл файл для записи.
2. `encoding="utf-8"` позволяет сохранять русские буквы.
3. `csv.writer(file)` создал объект для записи строк CSV.
4. `writerow()` записал заголовки таблицы.
5. `writerows()` записал несколько строк.
6. `with` автоматически закрыд фпйл после окончания блока.

## 2.2. Чтение CSV
Для чтения используется `csv.reader`.

Каждая строка при чтении представляется как список строк. Даже число `20` будет прочитано как текст. Если число нужно использовать в вычислениях, его нужно преобразовать через `int()` или `float()`.

````python
with open ('students.csv','r',encoding='utf-8', newline='') as file:
  reader=csv.reader(file)
  for row in reader:
    print(row)
````

## 2.3. `DictReader` - чтение таблицы как словаря
Если CSV содержит заголовки, удобно использовать `csv.DictReader`.

Тогда к столбцам можно обращаться по имени:


```
row['name']
row['age']
```
Это понятнее, чем запоминать индексы столбцов.


````python
with open('students.csv','r', encoding='utf-8', newline='') as file:
  reader=csv.DictReader(file)
  for row in reader:
    print(f'{row['name']} - {row['age']} лет, {row['city']}')
````

## 2.4. `DictWriter` - запись словарей в CSV
Если данные уже находятся в словарях, используем `csv.DictWriter`.

`fieldnames` задает названия столбцов

`writeheader()` записывает зоголовок

`writerows()` - записывает строки

````python
students=[
    {'name': 'Иван', 'age': 20, 'city': 'Москва'},
    {'name': 'Анна', 'age': 22, 'city': 'Казань'},
    {'name': 'Петр', 'age': 19, 'city': 'Минск'}
]
with open('students_dict.csv','w',encoding='utf-8',newline='') as file:
  fieldnames=['name','age','city']
  writer=csv.DictWriter(file,fieldnames)
  writer.writeheader()
  writer.writerows(students)
print('CSV со словарями создан')
````

##2.5. Разделитель `;`
CSV не обязан использовать запятую. Например, часто встречаются таблицы с разделителем `;`.

В таком случае указываем `delimiter=";"` и при записи, и при чтении.

````python
with open('products.csv','w', encoding='utf-8',newline='') as file:
  writer=csv.writer(file,delimiter=';')
  writer.writerow(['product','price','count'])
  writer.writerow(['Ноутбук', 8000, 2])
  writer.writerow(['Мышь', 2500, 10])
with open('products.csv', 'r', encoding='utf-8', newline='') as file:
  reader=csv.DictReader(file,delimiter=';')
  for row in reader:
    print(row)
````

### Основные инструменты `csv`
| Инструмент | Назначение |
|---|---|
|`csv.reader()`| читает строки CSV как списки |
|`csv.writer()`| записывает списки в CSV |
|`csv.DictReader()`| читает строки CSV как словари |
|`csv.DictWriter()`| записывает словари в CSV|
|`delimiter`| задает разделитель столбцов |

CSV хранит в себе табличные данные, но не хранит типы Python. Поэтому после чтения числа и другие значения при необходимости и преобразуются отедльно.

# Модуль `json`
JSON - текстовый формат для хранения структурированных данных. Нахвание происходит от **JavaScript Object Notation**, но формат используется не только в JavaScript.

Пример:


```
{
  'name': 'Иван',
  'age': 20,
  'skills': ['Python', 'SQL']
}
```

JSON удобен, когда данные имеют вложенную структуру: словари внутри списков, списки внутри словарей и т.д.

В Python JSON часто используется для конфигурационных файлав и обмена данными между программами, например при работе с API.

## 3.1. Python-объект ➡ JSON-строка
`json.dumps()` превращает объект Python в строку JSON.

Параметр `indent=4` добавляет отступы, чтобы JSON было удобно читать. `ensure_ascii=False` оставляет русские символы в обычном виде.

````python
import json
student={
    'name': 'Иван',
    'age': 20,
    'city': 'Москва',
    'skills': ['Python','SQL']
}
json_text=json.dumps(student, ensure_ascii=False, indent=4)
print(json_text)
````

## 3.2. JSON-строка ➡ Python
`json.loads()` выполняет обратную операцию.

После загрузки JSON-строки мы получаем обычный объект Python.

````python
json_text='{"name": "Анна", "age": 22, "active": true}'
data=json.loads(json_text)
print(data)
print(type(data))
print(data['name'])
````

###`loads()` и `dumps()`
Буква `s` связана со словом **string**:
- `json.dumps()` - Python ➡ JSON-строка
- `json.loads()` - JSON-строка ➡ Python

Эти функции работают именно со строками.

## 3.3. Запись JSON в файл
Для работы непосредственно с файлом используются `json.dump()` и `json.load()`.

Здесь буквы `s` нет, потому что мы не создаем промежуточную строку JSON.

````python
settings={
    'theme': 'dark',
    'language': 'ru',
    'notifications': True,
    'font_size': 16
}
with open('settings.json','w', encoding='utf-8') as file:
  json.dump(settings,file,ensure_ascii=False,indent=4)
print('settings.json сохранен')
````

## 3.4. Чтение JSON-файла
`json.load()` загружает данные непосредственно из файла.

````python
with open('settings.json','r', encoding='utf-8') as file:
  settings_from_file=json.load(file)
print(settings_from_file)
print('Тема:', settings_from_file['theme'])
print('размер шрифта:', settings_from_file['font_size'])
````

##3.5. Соответствие типов
|Python|JSON|
|---|---|
|`dict`|object|
|`list`,'tuple`|array|
|`str`|string|
|`int`,`float`|number|
|'True'/`False`|`true`/`false`|
|`None`|`null`|

Не каждый объект Python можно напрямую записать в JSON. Например, `set` и экземпляры пользовательских классов требуют предварительного преобразования.

# Модуль `zipfile`
ZIP-формат архива. Архив может содержать несколько файлов и папок в одном файле `.zip`.

Модуль `zipfile` позволяет из Python:
- создавать архивы
- добавлять файлы
- смотреть содержимое
- читать файлы прямо из архива
- распаковывать архивы

Например, программа может собрать CSV и JSON-отчет в один ZIP-файл.

## 4.1. Создание ZIP-архива
Откроем архим в режиме 'w' - запись.

Метож `write()` добавляет существующий файл.
`arcname` задает имя файла внутри архива.

````python
import zipfile
with zipfile.ZipFile('documents.zip','w', compression=zipfile.ZIP_DEFLATED) as archive:
  archive.write('students.csv', arcname='students.csv')
  archive.write('settings.json', arcname='settings.json')
print('Архив documents.zip создан')
````

### Что означает `Zip_DEFLATED`
`ZIP_DEFLATED` - стандартный вариант сжатия Deflate, поддерживаемый ZIP.

В результате архив обычно занимает меньше места, чем сумма исходных файлов, хотя степень сжатия зависит от содержимого.

##4.2. Просмотр содержимого архива
`namelist()` возвращает имена файлов, находящихся внутри ZIP.

````python
with zipfile.ZipFile('documents.zip','r') as archive:
  print(archive.namelist())
````

##4.3. Информация о файлах
`infolist()` возвращает обхекты `ZipInfo`. Они содержат имя файла, размер и другую информацию.

````python
with zipfile.ZipFile('documents.zip','r') as archive:
  for info in archive.infolist():
    print(f'{info.filename}:{info.file_size} ,байт')
````

## 4.4. Чтение файла прямо из архива
`read()` позволяет получить содержимое файла без предварительной распаковки на диск.

Результат `read()` имеет тип `bytes`, поэтому текстовый файл обычно нужно декодировать через `.decode('utf-8')`.

````python
with zipfile.ZipFile('documents.zip','r') as archive:
  data=archive.read('students.csv')
text=data.decode('utf-8')
print(text)
````

## 4.5. Распаковка
`extractall()` распаковывает содерживое архива в указанную папку.

Для архивов из неизвестных источников сначала стоит проверить `namelist()` и имена путей. Автоматическая распаковка непроверяемого архива в важную директорию может быть небезопасной.

````python
from pathlib import Path
extract_dir=Path('unzipped_documents')
extract_dir.mkdir(exist_ok=True)
with zipfile.ZipFile('documents.zip','r') as archive:
  archive.extractall(extract_dir)
print('Распаковка завершена')
print('Файлы:', [p.name for p in extract_dir.iterdir()])
````

# 5. Как связать `csv`, `json` и `zipfile`
Эти модули удобно использовать вместе.

Например:
**CSV ➡ обработка Python ➡ JSON ➡ ZIP**

Программа получает таблицу, обрабатывает данные, сохраняет результат в структурированном формате и затем собиарет несколько файлов в один архив.

````python
# 1. Читаем CSV
students = []

with open('students.csv', 'r', encoding='utf-8', newline='') as file:
    reader = csv.DictReader(file)
    for row in reader:
        row['age'] = int(row['age'])
        students.append(row)

# 2. Считаем средний возраст
average_age = sum(student['age'] for student in students) / len(students)

# 3. Сохраняем результат в JSON
report = {
    'students_count': len(students),
    'average_age': average_age,
    'students': students
}

with open('report.json', 'w', encoding='utf-8') as file:
    json.dump(report, file, ensure_ascii=False, indent=4)

# 4. Собираем CSV и JSON в ZIP
with zipfile.ZipFile('student_report.zip', 'w', compression=zipfile.ZIP_DEFLATED) as archive:
    archive.write('students.csv', arcname='students.csv')
    archive.write('report.json', arcname='report.json')

print(f'Количество студентов: {len(students)}')
print(f'Средний возраст: {average_age:.2f}')
print('Создан архив student_report.zip')
````
